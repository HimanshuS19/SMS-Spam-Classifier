# Word Clouds

```python
spam = " ".join(df[df.label==1]["clean_text"])

ham = " ".join(df[df.label==0]["clean_text"])

plt.figure(figsize=(15,6))

plt.subplot(121)

plt.imshow(
    WordCloud(background_color="black").generate(spam)
)

plt.axis("off")

plt.title("Spam Messages")

plt.subplot(122)

plt.imshow(
    WordCloud(background_color="white").generate(ham)
)

plt.axis("off")

plt.title("Ham Messages")

plt.show()
```
